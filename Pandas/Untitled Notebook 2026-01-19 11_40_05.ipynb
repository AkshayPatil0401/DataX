# **Null** **Handling**

```python
#Dropping Null from specific Subset

df1_modified = df1.dropna(subset=['name','department'])
```

```python
df1_modified
```

```python
df1
```

```python
df1_modified['email'] = df1_modified['email'].fillna('Unknown')
```

```python
display(df1_modified)
```

```python
df1_modified['salary']=df1_modified['salary'].fillna(df1_modified['salary'].mean())
display(df1_modified)
```

```python
df1_modified['join_date']=pd.to_datetime(df1_modified['join_date'], format='%Y-%m-%d',errors='coerce')
display(df1_modified)
```
